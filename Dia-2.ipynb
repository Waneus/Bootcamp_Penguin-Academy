###Challenge 🤺 - 20m

Escribe un programa que pregunte al usuario su edad para ingresar al festival Asunciónico y determine qué tipo de acceso puede tener.

Si el usuario tiene menos de 18 imprimir: "Acceso denegado — Solo zona Kids 🎈"

Si el usuario tiene mas de 18 pero menos de 21 imprimir: "Acceso permitido — No puede comprar bebidas alcoholicas 🎟️"

Si el usuario tiene mas de 21 imprimir: "Acceso total — Disfruta el festival 🎉"

```python
edad_usuario = int(input("Cual es su edad? "))
def posee_ticket():
    ticket_usuario = input("Posee ticket?")
    if ticket_usuario == "Si":
        print("Adelante, pase")

    elif ticket_usuario == "si":
        print("Adelante, pase")

    elif ticket_usuario == "no":
        print("Lo sentimos, no tiene ticket")

    elif ticket_usuario == "No":
        print("Lo sentimos, no tiene ticket")

    else:
        print("Lo sentimos no entendimos, vuelva a repetir por favor")

if edad_usuario < 18:
    print("Acceso Denegado - Solo zona Kids")

elif edad_usuario >= 22:
    print("Acceso total - Disfrute del ASuncionico!") 
    posee_ticket()

elif edad_usuario < 22:
    print("Acceso permitido - No puede comprar bebidas alcoholicas")
    posee_ticket()

else:
    print("No entendimos, vuelva a repetir por favor")
```

###Challenge 🤺 - 30m
Escribir un programa en el que se pregunte al usuario por una temperatura (en grados Celsius), imprima la temperatura escrita, y luego clasificar la temperatura ingresada en los siguientes niveles:

Entre -10 y 13 grados, que imprima "Ro'y koko".
Entre 14 y 23 grados, que imprima "Refresca'i".
Entre 24 y 28 grados, que imprima "Oikoite".
Entre 29 y 45 grados, que imprima "Hakurooooo".
Para todo lo demás, imprimir "Ñamano mba'e!".

```python
temperatura_usuario = float(input("Escriba su temperatura en grados Celsius🌡️: "))
print(f"{temperatura_usuario}°C")

if temperatura_usuario >= -10 and temperatura_usuario <= 13:
    print("Ro'y koko")

elif temperatura_usuario >= 14 and temperatura_usuario <= 23:
    print("Refresca'i")

elif temperatura_usuario >= 24 and temperatura_usuario <= 28:
    print("Oikoite")

elif temperatura_usuario >= 29 and temperatura_usuario <= 45:
    print("Hakurooooo")

else:
    print("Ñamano mba'e!")
```

###Escribir un programa que almacene una contraseña en una variable, y pregunte al usuario por la contraseña hasta que introduzca la contraseña correcta.

Cuando el usuario ingrese la contrasena correcta, imprimir un mensaje secreto.

```python
contrasena_secreta = "Grulla"

while True:
    contrasena_ingresada = input("Ingrese la contrasena secreta: ")
    if contrasena_secreta == contrasena_ingresada:
         print("Contrasena correcta")
         print("Mensaje secreto: Hola...")    
         break

    else:
        print(contrasena_ingresada)
        print("contrasena incorrecta")    
```

Challenge 🤺 - 15m
Solicitar al usuario que ingrese una frase.

Imprimir un mensaje cada vez que se encuentre una letra "o".

```python
frase_ingresada = input("Ingrese una frase")
contador = 0

for letra_frase in frase_ingresada:
    #print(letra_frase)
    if letra_frase == "O" or letra_frase == "o":
        print("Aqui hay una letra O ")
        contador += 1
print(f'Se encontraron {contador} letras o en total')
```

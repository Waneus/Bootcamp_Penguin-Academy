###Challenge
🎯 Objetivo: Practicar división y módulo.

Imagina que estás repartiendo caramelos: solicita al usuario ingresar el total de caramelos y la cantidad de niños, luego calcula cuántos caramelos recibe cada niño usando división entera y cuántos sobran usando módulo, e imprime ambos resultados.

```python
#variables pedidas
cant_total_caramelo = int(input("Cuántos caramelos posee en total?"))
cant_ninos = int(input("Cuántos niños hay en la fiesta?"))

#division entera
ninos_abastecidos = cant_total_caramelo // cant_ninos
print(f"{ninos_abastecidos} caramelo/s se repartieron para cada niño")

#resto
caramelos_sobrados = cant_total_caramelo % cant_ninos
print(f"Sobraron {caramelos_sobrados} caramelos post fiesta")
```

###Challenge
Crear una función de un proceso que se les ocurra que tenga varios pasos.

Ejecutar la función 3 veces.

```python
def gastos():

    nombre_gasto1 = input("Ingrese en que fue su primer gasto")
    valor_gasto1 = int(input("Ingrese el monto de su primer gasto"))

    nombre_gasto2 = input("Ingrese en que fue su primer gasto")
    valor_gasto2 = int(input("Ingrese el monto de su segundo gasto"))
    
    nombre_gasto3 = input("Ingrese en que fue su tercer gasto")
    valor_gasto3 = int(input("Ingrese el monto de su tercer gasto"))
    print(f"{nombre_gasto1}, {nombre_gasto2} y {nombre_gasto3} fue en lo que gasto")

    valor_total_gastado = valor_gasto1 + valor_gasto2 + valor_gasto3
    print(f"Este es el monto total que gasto {valor_total_gastado}")

gastos()
gastos()
gastos()
```

###Challenge 🤺 - 25m

Escribir una funcion que represente un sistema de pedidos de comida rapida.

La funcion debe recibir como parametros el nombre del producto y el tamaño del pedido.

Al ejecutarse, debe imprimir en pantalla el detalle del pedido de forma clara y ordenada.

Luego, llamar a la funcion pasando los argumentos correspondientes para simular que un cliente realiza un pedido.


```python
pedido_bebida = input("Cual sera su bebida?")
pedido_pasta = input("Diga la pasta que quiera consumir")
pedido_cantidad = int(input("Cuantos platos desea?"))

def pedido_productos(pasta, gaseosa, cantidad):
    print(f"Su {pasta} platos de {cantidad} se estan preparando!")
    print(f"Su {gaseosa} se le agrega...")
    print(f"Su {cantidad} plato de {pasta} con {gaseosa} esta listo para entregarse!")


pedido_productos(pedido_pasta, pedido_bebida, pedido_cantidad)
```
